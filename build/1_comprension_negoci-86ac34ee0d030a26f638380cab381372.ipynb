# 1. Comprensión del negocio

Previo al desarrollo de un análisis de datos, es imperativo realizar las siguientes acciones:

## 1.1 Definición del problema de negocio

El Centro de Salud Digital del Estado de Nayarit, es una entidad pública que opera unidades médicas regionales y gestiona el registro histórico de consultas de pacientes.

![](media/Nayarit.png)

El centro no puede predecir con certeza qué pacientes requerirán hospitalización antes de que lleguen al área de urgencias, lo que dificulta la asignación proactiva de camas y personal, generando ineficiencias operativas y posibles retrasos en la atención.

![](media/exceso-urgencias.png)

Para solucionar este problema, nos han contratado a nosotros, como Analistas de Datos Junior, para que, utilizando el registro histórico de consultas, limpiemos los datos, seleccionemos un modelo y algoritmo adecuado al caso, y ejecutemos un análisis predictivo que identifique los casos que requerirán hospitalización.

![](media/Prediction-Analytics.gif)

En resumen:

El **Centro de Salud Digital del Estado de Nayarit** busca **predecir** la necesidad de hospitalización de los pacientes antes de que ingresen al área de urgencias para **optimizar** proactivamente la asignación de camas y personal clínico.

![](media/recursos-hospital.jpg)

## 1.2 Evaluación del contexto

Para nuestro actual caso de estudio, contamos con los siguientes elementos:

### Fuente de datos disponible

Tenemos acceso a un **archivo histórico de consultas con 1010 registros de pacientes** que nos ha proporcionado el Centro de Salud Digital del Estado de Nayarit. Dicho archivo contiene 27 variables clínicas, socioeconómicas y de estilo de vida los pacientes, que fueron registradas durante su tratamiento, y ha sido almacenado en un formato `csv` que utiliza la codificación `UTF-8 BOM`.

![](media/CSV.gif)

### Objetivo de negocio

Nuestro objetivo durante el actual caso de estudio será **predecir qué pacientes requerirán hospitalización** antes de que lleguen a urgencias, para anticipar la asignación de camas y personal.

![](media/Predictive-Analytics-Transforming.gif)

## 1.3 Establecimiento de propósitos técnicos

Para este caso de estudio, utilizamos el lenguaje de programación `Python` junto con un conjunto de librerías especializadas:

- `pandas` y `numpy` para la manipulación y análisis de datos.
- `sklearn` y `mlxtend` para la implementación de modelos predictivos y técnicas de preprocesamiento.
- `featureranker` para la selección de características mediante un ensamble de métodos
- `matplotlib`, `seaborn` y `bokeh` para la visualización de datos.
