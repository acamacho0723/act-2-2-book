# 4. Modelado de datos

## 4.1 Técnicas de minería de datos (algoritmos predictivos)

La aplicación de técnicas de minería de datos en el ámbito de la salud permite transformar los registros clínicos históricos en conocimiento predictivo clínicamente accionable. Para el Centro de Salud Digital del Estado de Nayarit, la selección del paradigma algorítmico adecuado es un paso crítico, ya que define la estructura matemática con la que el modelo identificará los patrones de riesgo. Con el fin de fundamentar metodológicamente nuestra solución para la optimización de recursos y asignación de camas, es necesario delimitar las capacidades y alcances de los dos grandes enfoques del *Machine Learning*.

### Algoritmos de aprendizaje supervisado

Este enfoque se fundamenta en el aprendizaje a partir de ejemplos históricos previamente calificados o etiquetados. El algoritmo analiza un conjunto de datos de entrenamiento que incluye tanto las características de entrada como el resultado correcto asociado a cada una de ellas. A través de este proceso, el modelo aprende la función matemática o regla lógica que vincula las variables independientes con la variable dependiente. El objetivo final del aprendizaje supervisado es generalizar dicho conocimiento para realizar predicciones precisas o clasificaciones correctas al enfrentarse a nuevos datos que el sistema nunca antes ha procesado.

![](media/aprendizaje_supervisado.png)

Algunos ejemplos de estos algoritmos son:

- **Árbol de Decisión**
- **Naive Bayes Gaussiano**


### Algoritmos de aprendizaje no supervisado

A diferencia del esquema anterior, los algoritmos no supervisados operan sobre datos que carecen de una etiqueta o respuesta preestablecida. Su propósito principal es explorar la estructura intrínseca de los datos para descubrir agrupaciones naturales o patrones ocultos basados únicamente en las similitudes de los atributos. Si bien estas técnicas son valiosas para explorar el conjunto de datos o descubrir perfiles generales, no están diseñadas de forma nativa para resolver problemas de predicción guiada frente a un criterio crítico predeterminado.

![](media/aprendizaje_no_supervisado.png)

Algunos ejemplos de estos algoritmos son:

- **K-Means**
- **Apriori**


## 4.2 Selección de técnicas de modelado

### Árbol de Decisión

#### Podemos descomponer el funcionamiento de este algoritmo de Machine Learning en los siguientes elementos:

##### Concepto e Intuición del Modelo

![](media/DecisionTree/segmentos/segmento_01.gif)

Un árbol de decisión es una estructura binaria que fragmenta de manera recursiva el conjunto de datos. El objetivo principal es segmentar el espacio de atributos hasta alcanzar nodos hoja puros, es decir, agrupaciones que contengan registros pertenecientes a una única clase. El modelo se compone de nodos de decisión (donde se evalúan las condiciones) y nodos hoja (donde se dictamina la clasificación final).

##### Clasificación de Nuevas Instancias y Votación Mayoritaria

![](media/DecisionTree/segmentos/segmento_02.gif)

Para evaluar una nueva instancia, sus características atraviesan el árbol desde la raíz, comprobando las condiciones de cada nodo de decisión. Para encontrar las configuraciones que maximizan la eficiencia, el árbol de decisión busca las reglas que resultan en una mayor ganancia de información.

##### Selección de Divisiones Óptimas e Introducción a la Entropía

![](media/DecisionTree/segmentos/segmento_03.gif)

El árbol de decisión utiliza el aprendizaje automático para explorar los atributos y encontrar los umbrales correctos para dividir los datos de forma óptima. Para cuantificar la incertidumbre o la impureza en un nodo se utiliza la **Entropía**. Un nodo con una mezcla equitativa de clases posee la máxima entropía (valor de 1), mientras que un nodo puro de una sola clase posee una entropía de cero.

##### Ganancia de Información y Enfoque Codicioso

![](media/DecisionTree/segmentos/segmento_04.gif)

El modelo evalúa todas las divisiones posibles y selecciona la que maximiza la **Ganancia de Información** (la reducción de la entropía del nodo padre a los nodos hijos). Este algoritmo utiliza un enfoque codicioso, tomando la mejor decisión en el paso actual sin retroceder a modificar divisiones previas, lo que permite que el proceso de entrenamiento sea sumamente rápido y efectivo.

#### Y ahora ofrecemos una justificación al porque seleccionamos este algoritmo

Hemos seleccionado el **Árbol de Decisión** porque es un modelo interpretable, cuyas reglas de decisión pueden ser fácilmente comunicadas al personal médico del Centro de Salud Digital de Nayarit. Además, maneja bien interacciones entre variables y se adapta perfectamente a la naturaleza mixta de nuestros datos.

### Naive Bayes

#### Así como el Árbol de Decisión, hay una serie de elementos con los que podemos entender más fácilmente este algoritmo, como se muestra a continuación:

##### Fundamentos Probabilísticos y el Teorema de Bayes

![](media/NaiveBayes/segmentos/segmento_01.gif)

Para abordar nuestro problema de predicción de hospitalización desde una perspectiva probabilística, tratamos las características de los pacientes y la etiqueta objetivo (hospitalización) como variables aleatorias. El objetivo es encontrar la clase que maximice la probabilidad condicional (*posterior*) dados los síntomas o características del paciente. Dado que calcular esta probabilidad directamente es complejo, utilizamos el **Teorema de Bayes** , el cual nos permite invertir la condición utilizando la probabilidad previa (*prior*), la verosimilitud (*likelihood*) y la evidencia (*evidence*).

##### El Problema de la Combinación Exacta de Características

![](media/NaiveBayes/segmentos/segmento_02.gif)

Si intentáramos calcular las probabilidades buscando la combinación exacta de características de un paciente en el historial del Centro de Salud Digital de Nayarit, nos encontraríamos con que muchas combinaciones jamás han ocurrido previamente. Esto resultaría en probabilidades de cero, haciendo imposible comparar qué clase es la correcta.

##### La Suposición "Ingenua" (*Naive*) de Independencia Condicional

![](media/NaiveBayes/segmentos/segmento_03.gif)

Para solucionar el problema anterior, el algoritmo hace una suposición "ingenua" (*naive*): asume que todas las características del paciente son completamente independientes entre sí, dada la clase. Esto permite que la probabilidad conjunta se pueda calcular simplemente como el producto de las probabilidades individuales de cada atributo. Aunque en el mundo real médico los atributos a menudo están correlacionados, esta simplificación facilita enormemente los cálculos y suele producir resultados predictivos sumamente robustos.

##### Manejo de Variables Continuas

![](media/NaiveBayes/segmentos/segmento_04.gif)

Dado que en nuestro dataset tenemos variables continuas, no podemos simplemente contar frecuencias como lo haríamos con datos categóricos. Para el modelo **Gaussiano** , ajustamos una distribución normal conocida a nuestras características continuas.

#### Y ahora ofrecemos una justificación al porque seleccionamos este algoritmo

El **Naive Bayes Gaussiano** fue seleccionado por su simplicidad computacional y su buen desempeño en problemas clínicos con variables independientes. A pesar de la suposición "ingenua" de independencia entre atributos, este algoritmo es robusto frente a atributos irrelevantes, requiere poco entrenamiento y maneja de forma natural variables continuas mediante la asunción de una distribución normal. Su velocidad de inferencia lo hace adecuado para su potencial implementación en entornos de tiempo real en el Centro de Salud Digital de Nayarit.

### Reglas Apriori

#### El funcionamiento de las reglas Apriori se define a continuación

##### Concepto General del Aprendizaje de Reglas de Asociación

![](media/Apriori/segmentos/segmento1.gif)

El algoritmo analiza transacciones para identificar elementos que aparecen juntos en parejas o grupos.

##### Métrica de soporte

![](media/Apriori/segmentos/segmento2.gif)

Mide la frecuencia absoluta o popularidad de un elemento dentro de todo el conjunto de transacciones

##### Métrica de confianza

![](media/Apriori/segmentos/segmento3.gif)

Evalúa qué tan certera es una regla condicional.

##### Métrica de elevación

![](media/Apriori/segmentos/segmento4.gif)

Es el indicador final de la fuerza de la regla. Se calcula dividiendo la confianza entre el soporte y mide cuánto mejora nuestra predicción u oferta al usar el conocimiento previo (a priori) en comparación con hacer una recomendación completamente al azar a toda la población.

#### Seleccionamos este algoritmo como clasificador por el siguiente motivo

Seleccionamos este enfoque como algoritmo de comparación por su capacidad para generar reglas **interpretables** directamente desde los datos, lo que facilitaría la explicación de las predicciones al personal médico del Centro de Salud Digital de Nayarit. Además, las reglas de asociación pueden revelar interacciones inesperadas entre factores de riesgo que otros modelos de caja negra podrían ocultar. Aunque computacionalmente puede ser costoso en datasets de alta dimensionalidad, en nuestro contexto con un número moderado de características resulta viable y aporta una perspectiva complementaria a los clasificadores tradicionales (Árbol de Decisión y Naive Bayes).

```python
from mlxtend.frequent_patterns import apriori, association_rules
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from collections import Counter
import pandas as pd
import numpy as np
from IPython.display import display, HTML

X_over = df_over.drop('hospitalizacion', axis=1)
y_over = df_over['hospitalizacion']

X_train, X_test, y_train, y_test = train_test_split(X_over, y_over, test_size=0.3, random_state=42, stratify=y_over)

dt = DecisionTreeClassifier(max_depth=None, random_state=42)
dt.fit(X_train, y_train)

gnb = GaussianNB()
gnb.fit(X_train, y_train)

X_train_bin = X_train.astype(bool)
X_test_bin = X_test.astype(bool)

X_train_con_clase = X_train_bin.copy()
X_train_con_clase['hospitalizacion'] = y_train.values.astype(bool)

frequent_itemsets_clase = apriori(X_train_con_clase, min_support=0.25, use_colnames=True)
rules = association_rules(frequent_itemsets_clase, metric="confidence", min_threshold=0.6)
rules_clasif = rules[rules['consequents'].apply(lambda x: 'hospitalizacion' in x)]

X_test_bin_reset = X_test_bin.reset_index(drop=True)

def predict_apriori(X_test_bin, rules_clasif, default_class=0):
    y_pred = np.full(len(X_test_bin), default_class)
    for idx, row in X_test_bin.iterrows():
        for _, rule in rules_clasif.iterrows():
            antecedents = rule['antecedents']
            if antecedents.issubset(set(row[row].index.tolist())):
                if 'hospitalizacion' in rule['consequents']:
                    y_pred[idx] = 1
                else:
                    y_pred[idx] = 0
                break
    return y_pred

if len(rules_clasif) > 0:
    y_pred_apriori = predict_apriori(X_test_bin_reset, rules_clasif, default_class=0)
else:
    default = Counter(y_train).most_common(1)[0][0]
    y_pred_apriori = np.full(len(y_test), default)

f1_apriori = f1_score(y_test, y_pred_apriori)

display(HTML("<div style='text-align: center; font-family: Roboto, sans-serif;'>Entrenamiento exitoso de Árbol de Decisión, Naive Bayes y Reglas Apriori.</div>"))
```
